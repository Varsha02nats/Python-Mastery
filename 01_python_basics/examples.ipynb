#Track state across a single pass through a loop
the lowest price so far and the best profit so far(core of Best Time to Buy and Sell Stock problem)

```python
prices = [7,1,5,3,6,4]

min_price = float('inf')  #lowest price so far
best_profit = 0           #best profit so far

for p in prices:
    min_price = min(min_price, p)  #update the lowest price so far/ running minimum
    best_profit = max(best_profit, p - min_price) #update the best profit so far/ running maximum/ profit if we sold at price p today

print(best_profit) # Output: 5
```

#How it works:
1. Initialize min_price to positive infinity and best_profit to 0. Any real price will be less than positive infinity, so the first price we encounter will become the new min_price.
2. As we iterate through the list of prices, we update min_price to be the minimum of the current min_price and the current price p.
- This ensures that min_price always holds the lowest price we've seen so far.
3. We then calculate the potential profit if we were to sell at the current price p, which is p - min_price.
4. We update best_profit to be the maximum of the current best_profit and the potential profit we just calculated.
5. After iterating through all the prices, best_profit will hold the maximum profit we could have made by buying at the lowest price and selling at the highest price after that.

Simulataneous assignment allows you to assign values to multiple variables in a single line. This can be useful for swapping values or initializing multiple variables at once.

Simulataneous assignment evaluates the whole right-hand side of the assignment before performing the assignments. This means that the values are computed first, and then assigned to the variables in the order they appear on the left-hand side.

- for example, in the line `p, q = 1, 2`, the values `1` and `2` are evaluated first, and then assigned to `p` and `q` respectively. This allows for a concise and efficient way to assign multiple variables at once.

```python
p, q = 1, 2 

for _ in range(3):
    p, q = q, p+q    #right side uses old values of p and q to compute new values, then assigns them to p and q simultaneously
print("The value of p is:", p)  # Output: 5
print("The value of q is:", q)  # Output: 8
```
